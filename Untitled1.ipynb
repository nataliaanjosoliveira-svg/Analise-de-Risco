Célula 1: Instalação e Importação de Bibliotecas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from datetime import datetime, timedelta

# Reprodutibilidade
np.random.seed(42)

print("Bibliotecas carregadas com sucesso!")
```

2. GERAÇÃO DA BASE DE TRANSAÇÕES

```python
n = 1000

clientes = [f"CLI_{i:04d}" for i in range(1, 201)]

cidades = [
    "São Paulo",
    "Rio de Janeiro",
    "Belo Horizonte",
    "Curitiba",
    "Brasília",
    "Salvador",
    "Porto Alegre",
    "Recife"
]

dispositivos = [
    "Android",
    "iPhone",
    "Windows",
    "Mac"
]

status_transacao = [
    "Aprovada",
    "Recusada"
]

df = pd.DataFrame({
    "id_transacao": range(1, n + 1),
    "cliente": np.random.choice(clientes, n),
    "valor": np.round(np.random.exponential(scale=220, size=n) + 20, 2),
    "cidade": np.random.choice(cidades, n),
    "hora": np.random.randint(0, 24, n),
    "dispositivo": np.random.choice(dispositivos, n),
    "status": np.random.choice(
        status_transacao,
        n,
        p=[0.93, 0.07]
    )
})

df.head()
```

```python
# ============================================
# 3. VISÃO GERAL DA BASE
# ============================================

print(f"Quantidade de transações: {len(df)}")
print(f"Quantidade de clientes: {df['cliente'].nunique()}")
print(f"Valor total movimentado: R$ {df['valor'].sum():,.2f}")

display(df.head(10))
```

 4. CÁLCULO DO SCORE DE RISCO

```python

df["score_risco"] = 0

# Regra 1: valor da transação
df.loc[df["valor"] > 500, "score_risco"] += 20
df.loc[df["valor"] > 1000, "score_risco"] += 30

# Regra 2: horário da transação
df.loc[df["hora"].between(0, 5), "score_risco"] += 20

# Regra 3: transação recusada
df.loc[df["status"] == "Recusada", "score_risco"] += 20

# Classificação do risco
def classificar_risco(score):
    if score <= 30:
        return "Baixo"
    elif score <= 60:
        return "Médio"
    else:
        return "Alto"

df["nivel_risco"] = df["score_risco"].apply(classificar_risco)

display(df.head(10))
```

5. DISTRIBUIÇÃO DOS NÍVEIS DE RISCO

```python
distribuicao = (
    df["nivel_risco"]
    .value_counts()
    .reindex(["Baixo", "Médio", "Alto"])
    .fillna(0)
    .astype(int)
)

percentual = (
    df["nivel_risco"]
    .value_counts(normalize=True)
    .reindex(["Baixo", "Médio", "Alto"])
    .fillna(0)
    .mul(100)
    .round(2)
)

resultado_v2 = pd.DataFrame({
    "Quantidade": distribuicao,
    "Percentual": percentual
})

display(resultado_v2)

print(f"Valor total movimentado: R$ {df['valor'].sum():,.2f}")
```

 6. ANÁLISE DO COMPORTAMENTO FINANCEIRO

```python
# Calcula o valor médio movimentado por cada cliente
media_cliente = (
    df.groupby("cliente")["valor"]
      .mean()
      .reset_index()
      .rename(columns={"valor": "valor_medio_cliente"})
)

# Adiciona a média do cliente à base
df = df.merge(
    media_cliente,
    on="cliente",
    how="left"
)

# Calcula a diferença entre a transação e a média do cliente
df["desvio_valor"] = (
    df["valor"] - df["valor_medio_cliente"]
)

# Calcula o percentual de desvio
df["desvio_percentual"] = (
    df["desvio_valor"] /
    df["valor_medio_cliente"]
) * 100

display(
    df[
        [
            "cliente",
            "valor",
            "valor_medio_cliente",
            "desvio_valor",
            "desvio_percentual"
        ]
    ].head(10)
)
```

```python
# ============================================
# 7. MAIORES DESVIOS DE COMPORTAMENTO
# ============================================

maiores_desvios = (
    df.sort_values("desvio_percentual", ascending=False)
      [
          [
              "id_transacao",
              "cliente",
              "valor",
              "valor_medio_cliente",
              "desvio_percentual"
          ]
      ]
      .head(10)
)

display(maiores_desvios)
```

8. IDENTIFICAÇÃO DE TRANSAÇÕES ATÍPICAS

```python
df["comportamento_atipico"] = np.where(
    df["desvio_percentual"] > 100,
    "Sim",
    "Não"
)

print(
    "Quantidade de transações atípicas:",
    (df["comportamento_atipico"] == "Sim").sum()
)

display(
    df[
        [
            "id_transacao",
            "cliente",
            "valor",
            "valor_medio_cliente",
            "desvio_percentual",
            "comportamento_atipico"
        ]
    ]
    .query("comportamento_atipico == 'Sim'")
    .head(10)
)
```


# 9. IMPACTO FINANCEIRO DAS TRANSAÇÕES ATÍPICAS

```python
atipicas = df[
    df["comportamento_atipico"] == "Sim"
]

print(
    f"Transações atípicas: {len(atipicas)}"
)

print(
    f"Valor movimentado em transações atípicas: "
    f"R$ {atipicas['valor'].sum():,.2f}"
)

print(
    f"Valor médio das transações atípicas: "
    f"R$ {atipicas['valor'].mean():,.2f}"
)
```

 10. COMPORTAMENTO ATÍPICO X NÍVEL DE RISCO

```python
analise_comportamento = (
    df.groupby(
        ["comportamento_atipico", "nivel_risco"]
    )["id_transacao"]
    .count()
    .reset_index(name="quantidade")
)

display(analise_comportamento)
```

11. ANÁLISE DAS TRANSAÇÕES ATÍPICAS

```python
atipicas = df[df["comportamento_atipico"] == "Sim"]

risco_atipicas = (
    atipicas["nivel_risco"]
    .value_counts()
    .reindex(["Baixo", "Médio", "Alto"])
    .fillna(0)
    .astype(int)
)

print("Distribuição das transações atípicas:")
display(risco_atipicas)

print(
    f"Transações atípicas classificadas como Médio ou Alto risco: "
    f"{risco_atipicas['Médio'] + risco_atipicas['Alto']}"
)

print(
    f"Percentual: "
    f"{((risco_atipicas['Médio'] + risco_atipicas['Alto']) / len(atipicas) * 100):.2f}%"
)
```

 12. DISTRIBUIÇÃO DAS TRANSAÇÕES POR RISCO

```python
contagem_risco = (
    df["nivel_risco"]
    .value_counts()
    .reindex(["Baixo", "Médio", "Alto"])
    .fillna(0)
)

# Definindo cores semânticas para os níveis de risco
cores = ["#2ecc71", "#f39c12", "#e74c3c"]

plt.figure(figsize=(9, 5))
bars = plt.bar(contagem_risco.index, contagem_risco.values, color=cores, width=0.6)

# Adicionar o valor exato no topo de cada barra
for bar in bars:
    yval = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        yval + 10,
        f"{int(yval)} ({yval/len(df):.1%})",
        ha='center',
        va='bottom',
        fontweight='bold'
    )

plt.title("Distribuição das Transações por Nível de Risco", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Nível de Risco", fontsize=11)
plt.ylabel("Quantidade de Transações", fontsize=11)
plt.ylim(0, max(contagem_risco.values) * 1.15)  # Espaço para o texto no topo
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Remover bordas desnecessárias
for spine in ["top", "right"]:
    plt.gca().spines[spine].set_visible(False)

plt.tight_layout()
plt.show()
```

13. MODELO ISOLATION FOREST (DETECÇÃO AUTOMÁTICA)

```python
from sklearn.ensemble import IsolationForest

# Seleção de variáveis numéricas relevantes para o modelo
features = ["valor", "hora", "score_risco", "desvio_percentual"]

X = df[features].copy()

# Estimativa inicial da taxa de anomalias (contaminação) na base
# Usamos a taxa de transações atípicas prévia ou um valor fixo (ex: 5%)
taxa_contaminacao = 0.05

model_if = IsolationForest(
    n_estimators=100,
    contamination=taxa_contaminacao,
    random_state=42
)

# Ajuste do modelo e predição
# O Isolation Forest retorna -1 para anomalias e 1 para transações normais
df["pred_anomalia_if"] = model_if.fit_predict(X)

# Mapeando os rótulos para melhor leitura
df["is_anomalia_if"] = np.where(df["pred_anomalia_if"] == -1, "Anomalia", "Normal")

print(f"Total de Anomalias detectadas pelo Isolation Forest: {(df['is_anomalia_if'] == 'Anomalia').sum()}")
```

14. CONFRONTO: ISOLATION FOREST X SCORE DE RISCO

```python
matriz_comparativa = pd.crosstab(
    df["nivel_risco"],
    df["is_anomalia_if"],
    margins=True,
    margins_name="Total"
)

display(matriz_comparativa)

# Cruzamento com Comportamento Atípico vs Isolation Forest
matriz_comportamento = pd.crosstab(
    df["comportamento_atipico"],
    df["is_anomalia_if"],
    margins=True,
    margins_name="Total"
)

print("\n--- Confronto: Desvio de Comportamento (>100%) vs Isolation Forest ---")
display(matriz_comportamento)
```
